# 1. Introduction

[Go back to the "Table of contents"](#Table-of-contents)

The Antarctic Ice Sheet (AIS) is the largest block of ice on Earth. It is unique by its thickness, which averages 2 160 meters, and its surface of approximately 14 million square kilometers (an area larger than Europe), making it the biggest freshwater reservoir in the world. The ice sheet on the continent extends over the ocean and becomes ice shelf. This mass of ice plays a crucial role in the regulation of the Earth Climate, thanks to the relfective properties of the white ice covered areas surrounded by the dark sea surface, or the cold ocean surrounding the continent driving the oceaninc circulation around the globe. Its dynamics and evolution is a topic of significant concern for the scientific community.

Antarctica is particularly sensible to climate change, although there is a large climate variability over the region. The continent lies at the crossroads of numerous forcings : the ozone hole, disturbances caused by tropical storms and moonsoons, and of course the increases of greenhous gases in the atmosphere. This makes the Antartic Ice Sheet highly dynamic: climate records from the Antartic Peninsula show that  temperatures in this region rose by around 3.2°C in the second half of the 20th century, which is more than three times the global average. West Antartica has been melting at an accelerating rate, while the data on East Antarctica is less clear. In the end, as many unknowns remain about the processes governing the region and its potential feedback responses to climate change, there are still many uncertainties in the future contribution of the melting of the AIS to sea-level rise. According to the Fifth Assessment Report (AR5) from the Intergovernmental Panel on Climate Change (IPCC), even in the most extreme emissions scenario where surface mass gains in the East Antarctic Ice Sheet (EAIS) compensate for projected ice mass loss, the Antarctic Ice Sheet (AIS) is projected to contribute only a modest -0.08 to 0.14 meters to global sea level rise by the year 2100 ([Church et al., 2013](https://www.ipcc.ch/site/assets/uploads/2018/02/WG1AR5_Chapter13_FINAL.pdf)). On the other hand, in a high-temperature scenario, the expert assessment conducted by [Bamber et al. (2019)](https://doi.org/10.1073/pnas.1817205116) estimated a likely range of 0.03 to 0.46 meters from the West Antarctic Ice Sheet (WAIS) and a range of -0.04 to 0.11 meters from the East Antarctic Ice Sheet (EAIS) for the year 2100. These estimates are expected to increase to a range of 0.07 to 2.28 meters (WAIS) and -0.14 to 0.51 meters (EAIS) by the year 2300.

Monitoring the evolution of the AIS is therefore crucial to understand climate change and forecasts its future impacts across the globe. Measuring the mass of the Antarctic Ice sheet is however still a complex scientific task, and many methods exist including satellite altimetry, gravimetry, ice core analysis, ground-based GPS, airborne surveys, ice sheet models, etc.

## 4.1. First Look at the data

[Go back to the "Table of contents"](#Table-of-contents)

We have downloaded the needed data. The location of the file is : `data/C3S_GMB_GRACE_vers3.nc`. Let's now open it using the **xarray** library.

```python
dsi = xr.open_dataset('data/C3S_GMB_GRACE_vers3.nc')
```

```python
dsi
```

We see that the data has a few characteristics that make a little more difficult than anticipated to work with:
*  The time variable is not set as coordinate: this can lead to difficulties when performing calculations on the time series (data selection, resamplings, statistics, etc).
* The temporal resolution is irregular. Ideally, we would like to work with data with a regular timestep of exactly one month.

In order to solve this issue, we are going to re-create the same dataset, but this time we will specify the *time* variable as a coordinate, only select the variables related to the Antarctic, and resample the dataset to have a regular monthly timestep.


##  4.2. Pre-processing

[Go back to the "Table of contents"](#Table-of-contents)


```python
# re-generating the dataset at a convenient format

ds = xr.Dataset(
    coords=dict(
        time=dsi.time
    ),
    attrs=dsi.attrs,
)

for basin_nb in range(1,28):
    name = "AntIS_%d" % basin_nb
    if basin_nb in [25,26]:  # in the dataset we separate basins 25 and 26 for easier manipulation.
        ds[name] = dsi["AntIS_25_26"]   
        ds[name+"_er"] = dsi["AntIS_25_26_er"]
    else:
        ds[name] = dsi[name]
        ds[name+"_er"] = dsi[name+"_er"]


ds['East_AntIS_total'] = dsi['East_AntIS_total']
ds['East_AntIS_total_er'] = dsi['East_AntIS_total_er']

ds['West_AntIS_total'] = dsi['West_AntIS_total']
ds['West_AntIS_total_er'] = dsi['West_AntIS_total_er']

ds['AntPeninsula_total'] = dsi['AntPeninsula_total']
ds['AntPeninsula_total_er'] = dsi['AntPeninsula_total_er']

ds['AntIS_total'] = dsi['AntIS_total']
ds['AntIS_total_er'] = dsi['AntIS_total_er']

# resampling to have a regular time scale and filling holes with interpolation
ds = ds.resample(time='1M').mean().interpolate_na(dim='time')

ds
```

## 4.4. Plotting a simple graph

[Go back to the "Table of contents"](#Table-of-contents)


We are going to simply observe the global evolution of the Antartic Ice Sheet by plotting its mass balance as a function of time. The uncertainty on the ice mass change is also provided. 

```python
fig, ax = plt.subplots(figsize=(8,4))

dates = ds.time.astype('datetime64') # retrieving time coordinates
vals = ds.AntIS_total.data # ice mass balance
er = ds.AntIS_total_er.data # uncertainty

ax.plot(dates, vals, color='b')

# plotting the error margin
ax.fill_between(dates, vals-er, vals+er, facecolor='b', alpha=0.5)

ax.grid(True);
ax.set_ylabel('Gt')
ax.set_xlabel('year')
ax.set_title('Antarctic Ice Sheet Mass Anomaly over 2002-2022')
```

As expected, we see that the total mass of the Antarctic Ice Sheet has been decreasing for the past twenty years, and that there is also a seasonal pattern in the ice mass change. In the next sections we will look at the phenomenon in more detail. For now we can see in the product that there are Ice Mass Balance time series for the 3 major regions in Antarctica : East Antarctica, West Antarctica and the Antarctic Peninsula.

Before looking at the changes in these regions, let's smooth the seasonal Ice Mass variations by performing a yearly resampling of the data. We will do that only over the full years of data, so from 2003 to 2020.

```python
# yearly values

ds_y = ds.resample(time='1Y').mean().sel(time=slice('2003','2020'))
```

```python
# cumulative change

fig, ax = plt.subplots()

dates = ds_y.time.astype('datetime64')
vals_east = ds_y.East_AntIS_total.data - ds_y.East_AntIS_total.isel(time=[0]).item()
er_east = ds_y.East_AntIS_total_er.data

vals_west = ds_y.West_AntIS_total.data - ds_y.West_AntIS_total.isel(time=[0]).item()
er_west = ds_y.West_AntIS_total_er.data


vals_peninsula = ds_y.AntPeninsula_total.data - ds_y.AntPeninsula_total.isel(time=[0]).item()
er_p = ds_y.AntPeninsula_total_er.data


ax.plot(dates, vals_east, color='b', label='East Antarctic')
ax.fill_between(dates, vals_east-er_east, vals_east+er_east, facecolor='b', alpha=0.5)

ax.plot(dates, vals_west, color='r', label='West Antarctic')
ax.fill_between(dates, vals_west-er_west, vals_west+er_west, facecolor='r', alpha=0.5)

ax.plot(dates, vals_peninsula, color='y', label='Antarctic Peninsula')
ax.fill_between(dates, vals_peninsula-er_p, vals_peninsula+er_p, facecolor='y', alpha=0.5)

ax.set_title("Total Ice Mass evolution since 2003, by region.")
ax.set_ylabel('Gt')
ax.legend()
ax.grid(True)
```

This indicates that the Ice Loss is not equally distributed over the continent. We are going to look into more details which areas suffer the most from ice loss, and on the contrary which areas are actually accumulating ice over the studied period.

## 5.1. Computing the polygons

[Go back to the "Table of contents"](#Table-of-contents)


The `AntBasin` is indeed not a time series, it contains geographic information. It defines the contours of each Antarctic drainage basin (as defined by [Zwally et al](https://earth.gsfc.nasa.gov/cryo/data/polar-altimetry/antarctic-and-greenland-drainage-systems)) : each line in the table gives a **latitude**, a **longitude**, and the corresponding **basin number**. In the next cell we derive the 27 polygons from the coordinates given in the variable. This may take some time to execute !

```python
# creating polygons from coordnates:

# creating 27 empty lists.
basins_list = [[] for _ in range(27)]
for i in range(dsi['AntBasin'].data.shape[0]):
    # filling each list with the corresponding basin coordinates
    basins_list[int(dsi['AntBasin'].data[i,2]-1)].append(list(np.flip(dsi['AntBasin'].data[i,:][0:2])))
    

# computing polygons with geopandas
polygon_geom = []
for basin in range(len(basins_list)):
    if basin == 16: 
        # basin N°17 crosses the -180°/180° Longitude line.
        # so this one will be split in two to have a coherent visualisation
        poly1 = Polygon(basins_list[basin][39437:])
        poly2 = Polygon(basins_list[basin][:39437])

        multi_polygon = MultiPolygon([poly1, poly2])
        polygon_geom.append(multi_polygon)
    else:
        polygon_geom.append(Polygon(basins_list[basin]))
        
gdf = gpd.GeoDataFrame(index=range(1,28), crs='epsg:3857', geometry=polygon_geom)
gdf['basin_nb'] = range(1,28)

# coordinates of polygon representative points for annotations
gdf['coords'] = gdf['geometry'].apply(lambda x: x.representative_point().coords[:])
gdf['coords'] = [coords[0] for coords in gdf['coords']]
```

The result is a **GeoDataDrame**. It's a data table that also contain geospatial information. This dataframe will be very convenient in the next sections because we can add as many columns as we like. For now it only contains the polygons (***geometry***), their corresponding basin number (*basin_nb*) and the coordinates of the polygons centroids (*coords*).

## 5.2. Calculating trends and visualising Mass Change rate

[Go back to the "Table of contents"](#Table-of-contents)


The GRACE and GRACE-FO gravimetric products are particulary useful to calculate mass changes rates over long periods. We are going to perform linear regressions of the mass changes for each basins to see the spatial distribution of the ice mass change rates over the continent.

```python
# trend calculation:

# to ease trend calculation: manually specifying years
time = np.arange(2003, 2021)
vals = ds_y['AntIS_21'].data


slope, intercept = np.polyfit(time.reshape(-1,1).flatten(), vals, 1)
print(slope)

fig, ax =plt.subplots()
ax.plot(time, vals)

# plotting trend line
ax.plot(time, slope*time + intercept, color='b', linestyle='--')

ax.grid(True)
ax.set_xticks([2000,2005,2010,2015,2020]);

ax.set_ylabel("Gt")
ax.set_label("time")

ax.set_title("Example: Basin N°21 \nMass change rate: %.1f $Gt.y^{-1}$" %slope);
```

```python
# Calculation for every basin

time = np.arange(2003, 2021)
slope_list = []

for basin_nb in range(1,28):
    vals = ds_y["AntIS_%d" % basin_nb].data
    slope, intercept = np.polyfit(time.reshape(-1,1).flatten(), vals, 1)
    slope_list.append(slope)

gdf['yearly_mass_change_rate'] = slope_list
```

## 5.3. Analysing change rate acceleration

[Go back to the "Table of contents"](#Table-of-contents)


It is now clear that the drainage basins have very verying dynamics : most of West Antarctica suffers from very fast ice loss, whereas East Antarctica shows basins that are actually gaining ice, while a few other basins (N°12, 13 and 14) are also losing ice, but at a slower pace.

To deepen the analysis, we are now going to derive the timeseries in order to look at the trends followed by the ice mass change rates themselves. We can refer to those trends as the mass change acceleration.

```python
# same analysis, but with the derivative !

time = np.arange(2003, 2021)
slope_list = []

for basin_nb in range(1,28):
    vals = np.gradient(ds_y["AntIS_%d" % basin_nb].data)
    slope, intercept = np.polyfit(time.reshape(-1,1).flatten(), vals, 1)
    slope_list.append(slope)

gdf['yearly_mass_change_acc'] = slope_list

gdf.head()
```

```python
# Define the caracteristics of the plot
f = plt.figure(figsize=(10, 10))                                                    # create a figure and define its size
ax = plt.axes(projection=ccrs.SouthPolarStereo())                                       # create an ax and select the projection of the map
ax.coastlines()                                                                    # add the coastlines
gl = ax.gridlines(draw_labels=True)                        # add the longitude / latitude lines
gl.right_labels = False                                                            # remove latitude labels on the right
gl.top_labels = False                                                              # remove longitude labels on the top
ax.add_feature(cfeature.LAND, zorder=1, edgecolor='k')                             # add land mask
ax.set_extent([LON_MIN, LON_MAX, LAT_MIN, LAT_MAX],crs=ccrs.PlateCarree())         # define the extent of the map [lon_min,lon_max,lat_min,lat_max]

# Plot the sea total precipitation, set the min/max values of the colorbar and the colormap to use
im = gdf.plot(ax=ax, column='yearly_mass_change_acc', transform=ccrs.PlateCarree(), cmap='seismic_r', legend=True, vmin=-5, vmax=5, edgecolor='white', linewidth=0.5)
# Add the titles and legend
ax.set_title("Yearly Acceleration of Mass Change of the Antarctic Ice Shelf\nin $Gt.y^{-2}$")

for idx, row in gdf.iterrows():
    plt.annotate(text=row['basin_nb'], xy=row['coords'],
                 horizontalalignment='center', transform=ccrs.PlateCarree(), color='grey', fontsize=8)
    
# Save figure
plt.savefig('figs/GMB_acc.png')
```

We can see that the spatial variability of the ice mass change rates and the ice mass change accelerations are very different. For many of the basins that are gaining ice mass (ex: basins N°3, 6, 7, 17), the pace is actually slowing down over the years. Therefore, assuming the ice change rate keeps following a near-linear trend, there is a possiblity that some time in the future the ice change rate becomes negative and the drainage basins starts losing ice.

On the other hand, the basins in West Antartica that are already losing ice seem to follow an accelerating trend, showing that West Antartica is definitely the most fragile side of the Antarctic ice Sheet.

It is possible to subset a spatial dataset just like a timeseries dataset. The only difference is that this time, instead of the *time* coordinate, the selection constraints are applied to the *longitude* and/or *latitude* coordinates. Here is an example with the `slice` method.

```python
# Selecting only the Southern latitudes : 
# selecting only the years 2003 to 2020, for consistency with ice mass dataset
ds_s_south = ds_s.sel(latitude=slice(-90, -60)).sel(time=slice('2003', '2020'))

ds_s_south
```

Let's now calculate the correlation between the sea-level and the total Antarctic Ice-mass balance. We are going to use the **Pearson Coefficient**. It determines the linear correlation between two datasets. It varies between -1 and 1. A value of 1 implies that there is a perfectly linear relationship between the two distributions, and a value of -1 indicates a perfect inverse linear relationship. A value close to zero indicates that there is no linear pattern between the two datasets.

```python
# Correlation with Ice mass loss

# yearly values
ds_s_south_y = ds_s_south.resample(time='1Y').mean()

# to ease  calculation: manually specifying years
time = np.arange(2003, 2021)
vals_ice = ds_y['AntIS_total'].data
vals_sea = ds_s_south_y.mean(dim=['latitude', 'longitude'])['sla'].data

fig, ax = plt.subplots()
ax.scatter(vals_ice, vals_sea)
ax.grid(True)
ax.set_title("Scatter plot of SLA vs Ice mass change over 2003-2020.")
ax.set_xlabel("AIS mass change (GT)")
ax.set_ylabel("SLA (m)")

# correlation calculation using scipy
from scipy.stats import pearsonr
corr, _ = pearsonr(vals_ice, vals_sea)

print('Pearsons correlation: %.3f' % corr)
```

The Pearson coefficient of -0.727 indicates that there is a fairly strong inverse correlation between sea level and ice mass change. This makes sense, because as the ice melts (Ice mass change decreasing), the sea level increases. We can also see on the scatter plot that the points corresponding to the highest ice masses correspond to the lowest sea levels, and vice versa.

However, how is the sea level spatially distributed ? Let's emulate the basins from the Antarctic dataset by creating several sectors over the Antarctic ocean. We will create regular sectors with a 20° longitudinal length, covering from -90° to -60° latitude

For each sector, the sea levels will be spatially averaged, and the yearly trend computed.

```python
# creating a dataframe with regular sectors:
nb_sectors = 18
lon_slices = np.linspace(-180, 180, nb_sectors+1)
poly_list = []
y_series_list = []
trend_list = []
centroid_list = []




for i in range(nb_sectors):
    poly_list.append(Polygon([[lon_slices[i],-60], [lon_slices[i+1],-60], [lon_slices[i+1],-90], [lon_slices[i],-90]]))
    centroid_list.append((0.5*(lon_slices[i]+lon_slices[i+1]), -65)) # centroid for annotation
    
    tmp = ds_s_south_y.sel(longitude=slice(lon_slices[i], lon_slices[i+1]))
    ts = tmp.mean(dim=['latitude', 'longitude'])['sla']    # spatially averaging sea level over the sector
    y_series_list.append(ts.data)
    
    slope, intercept = np.polyfit(np.arange(len(ts.data)).reshape(-1,1).flatten(), ts.data, 1)   # computing yearly sea level trend
    trend_list.append(slope*1000)    # conversion to mm/year
    
    

gdf_sea = gpd.GeoDataFrame(index=range(nb_sectors), crs='epsg:3857', geometry=poly_list)
gdf_sea['sea_level_timeseries'] = y_series_list
gdf_sea['linear_trend_sea_level'] = trend_list
gdf_sea['sector_nb'] = range(1, nb_sectors+1)

gdf_sea['coords'] = centroid_list

gdf_sea
```

Let's now plot the sectors !

```python
# Define the caracteristics of the plot
f = plt.figure(figsize=(7, 7))                                                    # create a figure and define its size
ax = plt.axes(projection=ccrs.SouthPolarStereo())                                       # create an ax and select the projection of the map
ax.coastlines()                                                                    # add the coastlines
gl = ax.gridlines(draw_labels=True)                        # add the longitude / latitude lines
gl.right_labels = False                                                            # remove latitude labels on the right
gl.top_labels = False                                                              # remove longitude labels on the top
ax.add_feature(cfeature.LAND, zorder=2, edgecolor='k')                             # add land mask
ax.set_extent([LON_MIN, LON_MAX, LAT_MIN, LAT_MAX],crs=ccrs.PlateCarree())         # define the extent of the map [lon_min,lon_max,lat_min,lat_max]

# Plot the sea total precipitation, set the min/max values of the colorbar and the colormap to use
im = gdf_sea.plot(ax=ax, column='linear_trend_sea_level', transform=ccrs.PlateCarree(), cmap='BrBG', vmin=-3, vmax=3, edgecolor='k', linewidth=0.5, legend=True)
# Add the titles and legend
ax.set_title("Sea Level trends around the Antarctic \n between 2003-2020 in $mm.y^{-1}$")

for idx, row in gdf_sea.iterrows():
    plt.annotate(text=row['sector_nb'], xy=row['coords'],
                 horizontalalignment='center', transform=ccrs.PlateCarree(), color='b', fontsize=8)
    
```

We can see that the sea levels also have strong spatial variability! Most of the area around Antarctica has a rising trend going up to 3 mm/year, but near West Antarctica, the ocean follows a negative trend, which means the sea level is actually decreasing in this area!

This explained by the fact that there are more complex processes driving sea level evolution. The melting of the ice on the Antarctic creates gravitational adjustments that can make sea levels rise dramatically up in the Nothern Hemisphere. On the other hand, some areas near Antarctica can see their see level drop due to this gravitational pull.

Areas with decreasing sea levels remain few and sparse (see below). On average, the global sea level is going up about **3 millimeters per year**. Monitoring the waters fluxes, from glaciers to ocean, from ocean to rain, and so on, remains crucial to understand climate change better, improve the predictions and develop adequate mitigation solutions.
